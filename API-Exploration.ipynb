```python
import json 
import urllib.request

url = "https://api.erg.ic.ac.uk/AirQuality/Information/Species/Json"
with urllib.request.urlopen(url) as response: 
    raw = response.read()
jsondata = json.loads(raw)
print(raw)
print(json.loads(raw))
print(json.dumps (jsondata , indent=4))
print(len ([species for species in jsondata["AirQualitySpecies"]["Species"] if species["@SpeciesCode"]]))
# 6 species
```

```python
'''
An endpoint is a URL that returns JSON or XML data - we construct the endpoint from a 
mix of the base URL and the parameterss 
The set of parameters used to specify what kind of data the endpoint returns is 
Groupname , sitcode , speciescode , startdate , enddate , and the format of the data returned (Json or Xml)
Before calling an endpoint we know that the data returned will be specific to the set of parameters we input
'''
```

```python
import pandas as pd
tables = pd.read_html("https://api.erg.ic.ac.uk/AirQuality/help")
print(tables)
```

```python
print(len(tables)) # this returns 1 because theres only one table - containing lots of rows
print(tables[0].columns)
print(tables[0].head())
```

```python
dataframe = tables[0]
pd.set_option("display.max_colwidth", None) # prevents truncating of the output
dataframe.head()
```

```python
len(dataframe) # 70
dataframe["Uri"].str.endswith("/Json")
# df["Uri"] selects the Uri column (a Series, one string per row). The .str accessor
# lets us call a string method on every value, so .str.endswith("/Json") tests each
# Uri element-wise and returns a mask: a True/False Series of the same length.
# Indexing df with the mask, df[mask], keeps only the rows where it is True.
Mask = dataframe["Uri"].str.endswith("/Json")
JsonDataframe = dataframe[Mask]
print(JsonDataframe)
```

```python
print(len(JsonDataframe))
```

```python
from pathlib import Path
Path("snapshots").mkdir(exist_ok=True)
JsonDataframe.to_csv("snapshots/endpoints.csv", index=False)
```
